```python
def batch_embed(texts, batch_size=20):
    """
    Generate embeddings for a list of texts using OpenAI's
    text-embedding-3-small model.

    Parameters:
        texts: List of strings to embed
        batch_size: Number of texts sent per API request

    Returns:
        NumPy array of float32 embeddings
    """
    
    all_embeddings = []

    for start in range(0, len(texts), batch_size):
        batch = texts[start:start + batch_size]

        response = client.embeddings.create(
            model="text-embedding-3-small",
            input=batch
        )

        batch_embeddings = [
            item.embedding
            for item in response.data
        ]

        all_embeddings.extend(batch_embeddings)

        print(
            f"Embedded {min(start + batch_size, len(texts))}"
            f"/{len(texts)} documents"
        )

    return np.array(all_embeddings, dtype=np.float32)
```

```python
def embed_query(query):
    response = client.embeddings.create(
        model="text-embedding-3-small",
        input=[query]
    )

    return np.array(
        response.data[0].embedding,
        dtype=np.float32
    )


def semantic_search(query, top_k=5):
    query_vector = embed_query(query)

    distances, indices = index.search(
        query_vector.reshape(1, -1),
        top_k
    )

    results = []

    for distance, idx in zip(distances[0], indices[0]):
        results.append({
            "document_index": int(idx),
            "document": corpus[idx],
            "distance": float(distance)
        })

    return results
```

```python
import re


def keyword_search(query, corpus, top_k=5):
    query_words = set(
        re.findall(r"\b\w+\b", query.lower())
    )

    scored_documents = []

    for idx, document in enumerate(corpus):
        document_words = set(
            re.findall(r"\b\w+\b", document.lower())
        )

        overlap = query_words.intersection(document_words)

        score = len(overlap)

        scored_documents.append({
            "document_index": idx,
            "document": document,
            "score": score,
            "matched_words": sorted(overlap)
        })

    scored_documents.sort(
        key=lambda x: x["score"],
        reverse=True
    )

    return scored_documents[:top_k]
```

```python
import re
import pandas as pd


def keyword_search(query, corpus, top_k=5):
    query_words = set(
        re.findall(r"\b\w+\b", query.lower())
    )

    scored_documents = []

    for idx, document in enumerate(corpus):
        document_words = set(
            re.findall(r"\b\w+\b", document.lower())
        )

        overlap = query_words.intersection(document_words)

        scored_documents.append({
            "document_index": idx,
            "document": document,
            "score": len(overlap),
            "matched_words": sorted(overlap)
        })

    scored_documents.sort(
        key=lambda x: x["score"],
        reverse=True
    )

    return scored_documents[:top_k]
```

```python
def create_real_faiss_index(corpus):
    """
    Generate real OpenAI embeddings and build the final FAISS index.
    """

    embeddings = batch_embed(corpus)

    assert embeddings.dtype == np.float32
    assert embeddings.shape[0] == len(corpus)

    dimension = embeddings.shape[1]

    index = faiss.IndexFlatL2(dimension)
    index.add(embeddings)

    assert index.ntotal == len(corpus)

    return embeddings, index
```

```python
import re

stop_words = {
    "how", "can", "i", "do", "what", "is", "the",
    "a", "an", "to", "of", "in", "for", "and",
    "my", "from", "on", "with", "that", "this"
}

def keyword_search(query, corpus, top_k=5):
    query_words = set(re.findall(r"\b\w+\b", query.lower()))
    query_words = query_words - stop_words

    scored_documents = []

    for idx, document in enumerate(corpus):
        document_words = set(
            re.findall(r"\b\w+\b", document.lower())
        )

        overlap = query_words.intersection(document_words)

        scored_documents.append({
            "document_index": idx,
            "document": document,
            "score": len(overlap),
            "matched_words": sorted(overlap)
        })

    scored_documents.sort(
        key=lambda x: x["score"],
        reverse=True
    )

    return scored_documents[:top_k]
```
