```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

print("Embedding model loaded successfully!")
```

```python
embeddings = model.encode(sentences)

print("Number of sentences:", len(embeddings))
print("Embedding shape:", embeddings.shape)
```

```python
similarity_matrix = cosine_similarity(embeddings)

print("Cosine Similarity Matrix:")
print(similarity_matrix)
```

```python
similarity_df = pd.DataFrame(
    similarity_matrix,
    index=[f"S{i}" for i in range(1, 6)],
    columns=[f"S{i}" for i in range(1, 6)]
)

similarity_df
```

```python
pair_results = []

for i in range(len(sentences)):
    for j in range(i + 1, len(sentences)):
        score = similarity_matrix[i][j]
        
        pair_results.append({
            "Sentence 1": sentences[i],
            "Sentence 2": sentences[j],
            "Cosine Similarity": round(float(score), 4)
        })

pair_df = pd.DataFrame(pair_results)

pair_df
```

```python
highest_pair = pair_df.loc[
    pair_df["Cosine Similarity"].idxmax()
]

print("Most Similar Pair")
print("------------------")
print("Sentence 1:", highest_pair["Sentence 1"])
print("Sentence 2:", highest_pair["Sentence 2"])
print("Similarity Score:", highest_pair["Cosine Similarity"])
```

```python
ranked_pairs = pair_df.sort_values(
    by="Cosine Similarity",
    ascending=False
).reset_index(drop=True)

ranked_pairs
```

## Reflection

The two pizza-related sentences received the highest similarity score, which matched my intuition because both sentences express a similar meaning.

I was surprised that the embedding model can identify semantic similarity even when sentences do not use exactly the same words. This shows that embeddings capture meaning rather than simply counting words.

Compared with Bag-of-Words, embeddings provide a more useful representation when we want to compare the meaning of sentences.
