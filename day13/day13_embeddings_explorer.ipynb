```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

# 5 paraphrase pairs
pair_indices = [
    (0, 1),
    (5, 6),
    (10, 11),
    (15, 16),
    (17, 18)
]

results = []

for i, j in pair_indices:
    # TF-IDF similarity
    vectorizer = TfidfVectorizer()
    tfidf_vectors = vectorizer.fit_transform([sentences[i], sentences[j]])
    tfidf_score = cosine_similarity(
        tfidf_vectors[0:1],
        tfidf_vectors[1:2]
    )[0][0]

    # Embedding similarity
    embedding_score = cosine_similarity(
        embeddings[i:i+1],
        embeddings[j:j+1]
    )[0][0]

    results.append({
        "Sentence A": sentences[i],
        "Sentence B": sentences[j],
        "Topic": topics[i],
        "TF-IDF Similarity": round(float(tfidf_score), 4),
        "Embedding Similarity": round(float(embedding_score), 4)
    })

comparison_df = pd.DataFrame(results)

comparison_df
```

```python
# Identify pairs where TF-IDF is relatively low
# but embedding similarity is relatively high

comparison_df["Semantic Gap"] = (
    comparison_df["Embedding Similarity"]
    - comparison_df["TF-IDF Similarity"]
)

paraphrase_candidates = comparison_df.sort_values(
    "Semantic Gap",
    ascending=False
).head(5)

print("Top 5 paraphrase candidates:")
paraphrase_candidates[
    [
        "Topic",
        "Sentence A",
        "Sentence B",
        "TF-IDF Similarity",
        "Embedding Similarity",
        "Semantic Gap"
    ]
]
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

def embed_and_recommend(query_sentence, corpus_sentences, top_k=3):
    """
    Return top-k semantically similar sentences
    using dense sentence embeddings.
    """
    
    # Generate embeddings for query + corpus
    all_texts = [query_sentence] + list(corpus_sentences)

    all_embeddings = model.encode(
        all_texts,
        convert_to_numpy=True
    )

    query_embedding = all_embeddings[0].reshape(1, -1)
    corpus_embeddings = all_embeddings[1:]

    # Calculate cosine similarity
    scores = cosine_similarity(
        query_embedding,
        corpus_embeddings
    )[0]

    # Get highest similarity scores
    top_indices = np.argsort(scores)[::-1][:top_k]

    recommendations = []

    for idx in top_indices:
        recommendations.append({
            "sentence": corpus_sentences[idx],
            "similarity": round(float(scores[idx]), 4)
        })

    return recommendations
```

```python
query = "A soccer team won an exciting championship match."

recommendations = embed_and_recommend(
    query,
    sentences,
    top_k=3
)

for rank, item in enumerate(recommendations, start=1):
    print(f"{rank}. {item['sentence']}")
    print(f"   Similarity: {item['similarity']}")
```

```python
from sklearn.cluster import KMeans
import pandas as pd

kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans.fit_predict(embeddings)

cluster_df = pd.DataFrame({
    "Sentence": sentences,
    "Original Topic": topics,
    "Cluster": cluster_labels
})

cluster_df
```

```python
cluster_topic_table = pd.crosstab(
    cluster_df["Cluster"],
    cluster_df["Original Topic"]
)

print("Cluster vs Original Topic:")
cluster_topic_table
```

```python
# Assign each cluster its majority topic
cluster_topic_map = (
    cluster_df.groupby("Cluster")["Original Topic"]
    .agg(lambda x: x.mode()[0])
    .to_dict()
)

cluster_df["Predicted Topic"] = cluster_df["Cluster"].map(cluster_topic_map)

accuracy = (
    cluster_df["Original Topic"] == cluster_df["Predicted Topic"]
).mean()

print(f"Cluster-topic alignment: {accuracy:.2%}")
```

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

pca = PCA(n_components=2, random_state=42)

embedding_2d = pca.fit_transform(embeddings)

plot_df = pd.DataFrame({
    "x": embedding_2d[:, 0],
    "y": embedding_2d[:, 1],
    "topic": topics,
    "sentence": sentences
})

plt.figure(figsize=(10, 7))

for topic in plot_df["topic"].unique():
    subset = plot_df[plot_df["topic"] == topic]
    
    plt.scatter(
        subset["x"],
        subset["y"],
        label=topic,
        s=80
    )

    for _, row in subset.iterrows():
        plt.annotate(
            topic,
            (row["x"], row["y"]),
            xytext=(5, 5),
            textcoords="offset points"
        )

plt.title("2D PCA Projection of Sentence Embeddings")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## Sparse TF-IDF vs Dense Embeddings

### TF-IDF: Sparse Representation

TF-IDF represents a sentence using a vector based on the words
appearing in the corpus.

Each dimension corresponds to a vocabulary term. Most dimensions
are zero, so the representation is called sparse.

TF-IDF is especially useful for:
- Exact keyword matching
- Lexical similarity
- Search where specific terms matter
- Interpretable word-based retrieval

However, TF-IDF does not naturally understand synonyms or paraphrases.

For example:

"The football team won the championship."

and

"The soccer club captured the league title."

have limited word overlap even though they have a very similar meaning.

### Embeddings: Dense Representation

Sentence embeddings represent text as dense numerical vectors in a
continuous semantic space.

Each sentence is represented by many non-zero dimensions. Sentences
with similar meanings tend to have vectors that are close together.

Embeddings are useful for:
- Semantic search
- Recommendation systems
- Paraphrase detection
- Document clustering
- Retrieval-Augmented Generation (RAG)

### Key Difference

TF-IDF primarily captures **word overlap**.

Embeddings primarily capture **semantic meaning**.

Therefore, embeddings generally perform better when two sentences express
similar ideas using different vocabulary.

## Conclusion

This experiment demonstrates the difference between lexical and
semantic representations of text.

TF-IDF relies heavily on vocabulary overlap, while dense embeddings
place semantically related sentences closer together in vector space.

The paraphrase comparison showed that sentences with different wording
can have relatively low TF-IDF similarity while maintaining higher
embedding similarity.

The semantic recommendation function successfully retrieved sentences
based on meaning rather than simple keyword matching.

K-Means clustering was then applied to the sentence embeddings. The
resulting clusters showed alignment with the four manually assigned
topics: Sports, Technology, Cooking, and Travel.

PCA provided a 2D visualization of the embedding space and helped
inspect the separation between topics.
