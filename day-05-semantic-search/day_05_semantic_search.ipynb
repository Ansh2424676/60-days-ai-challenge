```python
# Day 5 - Build a Mini Search Engine

sentences = [
    "Artificial intelligence is changing the way people work.",
    "Machine learning allows computers to learn from data.",
    "Deep learning uses neural networks to solve complex problems.",
    "Natural language processing helps computers understand human language.",
    "Semantic search finds information based on meaning.",
    "Keyword search looks for exact words in a document.",
    "Vector databases store numerical representations of information.",
    "Embeddings convert text into numerical vectors.",
    "Retrieval augmented generation improves answers using external knowledge.",
    "Recommendation systems suggest relevant products to users.",
    "Computer vision enables machines to understand images and videos.",
    "Large language models can generate human-like text.",
    "Chatbots use artificial intelligence to communicate with users.",
    "Data preprocessing prepares raw data for machine learning models.",
    "Cosine similarity measures how similar two vectors are.",
    "Search engines help users find relevant information quickly.",
    "Neural networks are inspired by the structure of the human brain.",
    "AI assistants can answer questions and perform useful tasks.",
    "Semantic similarity focuses on the meaning of sentences.",
    "Modern AI applications combine search, embeddings, and language models."
]

print(f"Number of sentences: {len(sentences)}")
```

```python
from sentence_transformers import SentenceTransformer

# Load a lightweight sentence embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")

# Convert all sentences into vectors
embeddings = model.encode(
    sentences,
    convert_to_numpy=True
)

print("Embedding shape:", embeddings.shape)
```

```python
print("Total sentences:", len(sentences))
print("Total embeddings:", len(embeddings))
print("Vector dimensions:", embeddings.shape[1])
```

```python
from sklearn.metrics.pairwise import cosine_similarity


def semantic_search(query, top_k=3):
    """
    Find the most semantically similar sentences
    for a given query.
    """

    # Convert query into an embedding
    query_embedding = model.encode(
        [query],
        convert_to_numpy=True
    )

    # Calculate similarity between query and all sentences
    similarities = cosine_similarity(
        query_embedding,
        embeddings
    )[0]

    # Get indices of top results
    top_indices = similarities.argsort()[-top_k:][::-1]

    results = []

    for index in top_indices:
        results.append({
            "sentence": sentences[index],
            "score": float(similarities[index])
        })

    return results
```

```python
query = "How do computers understand human language?"

results = semantic_search(query)

for i, result in enumerate(results, start=1):
    print(f"{i}. {result['sentence']}")
    print(f"   Similarity: {result['score']:.4f}")
    print()
```

```python
def keyword_search(query, top_k=3):
    query_words = set(query.lower().split())

    results = []

    for sentence in sentences:
        sentence_words = set(sentence.lower().split())

        score = len(query_words & sentence_words)

        results.append({
            "sentence": sentence,
            "score": score
        })

    results.sort(
        key=lambda x: x["score"],
        reverse=True
    )

    return results[:top_k]


print("Keyword search function is ready!")
```

# Semantic Search vs Keyword Search

## Conclusion

This mini search engine demonstrates the difference between
traditional keyword search and semantic search.

### Keyword Search
- Finds results using exact word overlap.
- Simple and fast.
- May miss relevant results when different words have the same meaning.

### Semantic Search
- Converts sentences and queries into embeddings.
- Uses cosine similarity to compare meaning.
- Can retrieve relevant information even when the wording is different.

### Pipeline

Text Dataset
→ Sentence Embeddings
→ Query Embedding
→ Cosine Similarity
→ Top 3 Semantic Results

This approach is a foundation for modern AI applications such as
RAG, semantic recommendations, and AI-powered search.
