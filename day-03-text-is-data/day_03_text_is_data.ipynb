# Day 3 — Text is Data: Turning Words into Numbers

## Objective

In this notebook, I will:

1. Tokenize a paragraph using NLTK.
2. Remove stop words and punctuation.
3. Create a preprocessed text dataset.
4. Convert text into a Bag-of-Words representation.

### NLP Pipeline

Raw Text → Tokenization → Cleaning → Bag of Words → Numerical Data

```python
import nltk
import string
import pandas as pd

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import CountVectorizer

print("Libraries imported successfully!")
```

```python
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")

print("NLTK resources downloaded successfully!")
```

```python
text = """
Artificial Intelligence is changing the world.
AI systems can understand text, process information, and help people solve problems.
Natural Language Processing allows machines to work with human language.
"""

print("Original Text:")
print(text)
```

```python
tokens = word_tokenize(text)

print("Tokenized Text:")
print(tokens)
```

```python
lower_tokens = [token.lower() for token in tokens]

print("Lowercase Tokens:")
print(lower_tokens)
```

```python
stop_words = set(stopwords.words("english"))

filtered_tokens = [
    token for token in lower_tokens
    if token not in stop_words
]

print("After Stop Word Removal:")
print(filtered_tokens)
```

```python
clean_tokens = [
    token for token in filtered_tokens
    if token not in string.punctuation
]

print("After Removing Punctuation:")
print(clean_tokens)
```

```python
preprocessed_text = " ".join(clean_tokens)

print("Final Preprocessed Text:")
print(preprocessed_text)
```

```python
data = {
    "original_text": [
        "AI is changing the world.",
        "Natural Language Processing helps machines understand human language.",
        "Machine learning can process large amounts of text."
    ]
}

df = pd.DataFrame(data)

df
```

```python
def preprocess_text(text):
    tokens = word_tokenize(text.lower())
    
    tokens = [
        token for token in tokens
        if token not in stop_words
        and token not in string.punctuation
    ]
    
    return " ".join(tokens)
```

```python
df["preprocessed_text"] = df["original_text"].apply(preprocess_text)

df
```

```python
vectorizer = CountVectorizer()

bow_matrix = vectorizer.fit_transform(df["preprocessed_text"])

print("Bag-of-Words created successfully!")
```

```python
vocabulary = vectorizer.get_feature_names_out()

print("Vocabulary:")
print(vocabulary)
```

```python
bow_array = bow_matrix.toarray()

print("Bag-of-Words Matrix:")
print(bow_array)
```

```python
bow_df = pd.DataFrame(
    bow_array,
    columns=vocabulary
)

bow_df
```

```python
sample = "AI is changing the world!"

print("1. Original Text:")
print(sample)

tokens = word_tokenize(sample)

print("\n2. Tokenization:")
print(tokens)

lower_tokens = [token.lower() for token in tokens]

print("\n3. Lowercase:")
print(lower_tokens)

without_stopwords = [
    token for token in lower_tokens
    if token not in stop_words
]

print("\n4. Stop Words Removed:")
print(without_stopwords)

final_tokens = [
    token for token in without_stopwords
    if token not in string.punctuation
]

print("\n5. Punctuation Removed:")
print(final_tokens)

print("\n6. Final Preprocessed Text:")
print(" ".join(final_tokens))
```

## Conclusion

In this task, raw text was converted into a form that can be used by machine learning algorithms.

The preprocessing pipeline was:

Raw Text
→ Tokenization
→ Lowercase Conversion
→ Stop Word Removal
→ Punctuation Removal
→ Bag of Words
→ Numerical Representation

The final Bag-of-Words matrix represents each document using numerical word counts.
