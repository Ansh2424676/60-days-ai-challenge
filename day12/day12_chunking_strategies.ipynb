```python
# Manual fixed-size chunking
def fixed_size_chunking(text, chunk_size):
    chunks = []

    for start in range(0, len(text), chunk_size):
        chunk = text[start:start + chunk_size]
        chunks.append(chunk)

    return chunks
```

```python
def recursive_chunking(text, chunk_size, chunk_overlap):
    splitter = RecursiveCharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap,
        separators=["\n\n", "\n", ". ", " ", ""]
    )
    return splitter.split_text(text)
```

```python
def find_boundary_failures(text, chunk_size, max_examples=3):
    failures = []

    # Check every fixed-size boundary
    for boundary in range(chunk_size, len(text), chunk_size):

        # Character immediately before and after boundary
        before = text[max(0, boundary - 100):boundary]
        after = text[boundary:boundary + 100]

        # Find sentence punctuation around the boundary
        last_period = before.rfind(".")
        next_period = after.find(".")

        # If there is no sentence-ending period near the boundary,
        # the boundary may have split a sentence.
        if last_period == -1 or next_period == -1:
            
            failures.append({
                "character_position": boundary,
                "before_boundary": before,
                "after_boundary": after
            })

        if len(failures) >= max_examples:
            break

    return failures
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

def retrieve_best_chunk(chunks, query):
    corpus = chunks + [query]

    vectorizer = TfidfVectorizer(
        stop_words="english"
    )

    vectors = vectorizer.fit_transform(corpus)

    chunk_vectors = vectors[:-1]
    query_vector = vectors[-1]

    similarities = cosine_similarity(
        query_vector,
        chunk_vectors
    ).flatten()

    best_index = similarities.argmax()

    return {
        "chunk_index": int(best_index),
        "score": float(similarities[best_index]),
        "chunk": chunks[best_index]
    }
```

```python
retrieval_query = (
    "What are the goals, capabilities and applications "
    "of artificial intelligence?"
)

print("Retrieval query:")
print(retrieval_query)
```

```python
chunk_sizes = [100, 300, 500, 1000]
overlaps = [0, 50, 100]

experiment_results = []

for chunk_size in chunk_sizes:
    for overlap in overlaps:

        try:
            chunks = recursive_chunking(
                document_text,
                chunk_size=chunk_size,
                chunk_overlap=overlap
            )

            result = retrieve_best_chunk(
                chunks,
                retrieval_query
            )

            experiment_results.append({
                "chunk_size": chunk_size,
                "overlap": overlap,
                "num_chunks": len(chunks),
                "retrieval_score": result["score"],
                "retrieved_chunk": result["chunk"],
                "status": "success"
            })

        except Exception as e:
            experiment_results.append({
                "chunk_size": chunk_size,
                "overlap": overlap,
                "num_chunks": None,
                "retrieval_score": None,
                "retrieved_chunk": str(e),
                "status": "error"
            })

results_df = pd.DataFrame(experiment_results)

print("Total configurations:", len(results_df))

results_df[
    [
        "chunk_size",
        "overlap",
        "num_chunks",
        "retrieval_score",
        "status"
    ]
]
```

```python
important_terms = [
    "intelligence",
    "goal",
    "capabilities",
    "applications",
    "learning",
    "reasoning",
    "problem",
    "knowledge"
]

def contextual_completeness(chunk):
    text = chunk.lower()

    matched_terms = [
        term
        for term in important_terms
        if term in text
    ]

    score = len(matched_terms) / len(important_terms)

    return score, matched_terms
```

```python
results_df["context_completeness"] = 0.0
results_df["quality_score"] = 0.0

for index, row in results_df.iterrows():

    if row["status"] != "success":
        continue

    completeness, matched_terms = contextual_completeness(
        row["retrieved_chunk"]
    )

    results_df.loc[index, "context_completeness"] = completeness

    results_df.loc[index, "quality_score"] = (
        row["retrieval_score"] * 0.6
        + completeness * 0.4
    )

results_df[
    [
        "chunk_size",
        "overlap",
        "num_chunks",
        "retrieval_score",
        "context_completeness",
        "quality_score",
        "status"
    ]
]
```

```python
successful_results = results_df[
    results_df["status"] == "success"
].copy()

successful_results = successful_results.sort_values(
    by="quality_score",
    ascending=False
).reset_index(drop=True)

best_result = successful_results.iloc[0]

print("=" * 70)
print("BEST CHUNKING CONFIGURATION")
print("=" * 70)

print("Chunk Size:", int(best_result["chunk_size"]))
print("Overlap:", int(best_result["overlap"]))
print("Number of Chunks:", int(best_result["num_chunks"]))
print("Retrieval Score:", round(best_result["retrieval_score"], 4))
print("Context Completeness:", round(best_result["context_completeness"], 4))
print("Overall Quality Score:", round(best_result["quality_score"], 4))

print("\nRetrieved Context:")
print("-" * 70)
print(best_result["retrieved_chunk"])
```

# Final Chunking Strategy Recommendation

## Experimental Findings

This experiment compared manual fixed-size chunking with LangChain's
RecursiveCharacterTextSplitter on a long Wikipedia article about Artificial
Intelligence.

Four chunk sizes were evaluated:

- 100 characters
- 300 characters
- 500 characters
- 1000 characters

Three overlap values were evaluated:

- 0 characters
- 50 characters
- 100 characters

This resulted in 12 chunking configurations.

The same retrieval query was used for every configuration:

> What are the goals, capabilities and applications of artificial intelligence?

## Results

The experiment showed that very small chunks provide highly focused text but
can lose important surrounding context. Larger chunks preserve more context,
but they may also introduce irrelevant information into the retrieved result.

Chunk overlap helps preserve continuity when an important sentence or concept
falls close to a chunk boundary.

The best-performing configuration was selected using a combined quality score
based on retrieval similarity and contextual completeness.

## Recommended Strategy

For this document, the recommended configuration is the configuration with
the highest overall quality score reported by the experiment.

The general recommendation for similar long-form documents is to start with:

- Chunk size: 500 characters
- Chunk overlap: 50-100 characters
- Strategy: RecursiveCharacterTextSplitter

This provides a useful balance between context preservation and retrieval
precision.

## Boundary Failure Analysis

Manual fixed-size chunking demonstrated that arbitrary character boundaries
can split sentences in the middle of a thought.

Three such boundary failures were identified and documented with their exact
character positions in the experiment.

These failures demonstrate why semantic-aware recursive splitting is generally
preferable to naive fixed-size splitting for Retrieval-Augmented Generation
(RAG) systems.

## Conclusion

Chunking is not simply a preprocessing step. It directly affects retrieval
quality.

The experiment demonstrates that:

1. Very small chunks can lose context.
2. Very large chunks can contain unnecessary information.
3. Overlap helps maintain continuity.
4. Recursive splitting produces more natural boundaries.
5. Chunk size should be selected based on document structure and retrieval
   requirements.
