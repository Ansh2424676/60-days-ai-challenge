```python
# Day 8 — How Machines Read Human Language
# Text Preprocessing using NLTK and Regex

import re
import nltk

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download required NLTK resources
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")

print("NLTK setup complete!")
```

```python
# Three noisy real-world text samples

samples = {
    "Social Media": "WOW!!! This phone is AMAZING 😍😍 I can't believe how FAST it is!!! #BestPhone",
    
    "News Article": "The government announced new policies today, saying that the changes will improve digital services and public access.",
    
    "Customer Review": "I bought this product last week... The quality is GOOD, but the delivery was VERY slow 😒."
}

for source, text in samples.items():
    print(f"\n--- {source} ---")
    print(text)
```

```python
# Tokenization: split text into words and sentences

for source, text in samples.items():
    print(f"\n{'='*50}")
    print(source)
    print(f"{'='*50}")

    words = word_tokenize(text)
    sentences = sent_tokenize(text)

    print("\nWord Tokens:")
    print(words)

    print("\nSentence Tokens:")
    print(sentences)
```

```python
# Remove stopwords and compare vocabulary size

stop_words = set(stopwords.words("english"))

for source, text in samples.items():
    words = word_tokenize(text.lower())

    # Vocabulary before removing stopwords
    vocab_before = set(words)

    # Remove stopwords
    filtered_words = [
        word for word in words
        if word.isalpha() and word not in stop_words
    ]

    vocab_after = set(filtered_words)

    print(f"\n--- {source} ---")
    print("Before:", words)
    print("After :", filtered_words)
    print("Vocabulary before:", len(vocab_before))
    print("Vocabulary after :", len(vocab_after))
```

```python
# Remove punctuation using Regex.
# This is necessary because punctuation can create unnecessary tokens
# and noise for downstream NLP processing.

for source, text in samples.items():
    cleaned_text = re.sub(r"[^\w\s]", "", text)

    print(f"\n--- {source} ---")
    print("Before:", text)
    print("After :", cleaned_text)
```

```python
# Stemming vs Lemmatization

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

words = ["playing", "played", "plays", "studies", "studying", "better", "cars"]

print(f"{'Word':<15} {'Stem':<15} {'Lemma':<15}")
print("-" * 45)

for word in words:
    stem = stemmer.stem(word)
    lemma = lemmatizer.lemmatize(word)
    print(f"{word:<15} {stem:<15} {lemma:<15}")
```

```python
# Complete before-and-after comparison

for source, text in samples.items():
    print("\n" + "=" * 70)
    print(source)
    print("=" * 70)

    # 1. Raw text
    print("\n1. RAW:")
    print(text)

    # 2. Lowercase + Word Tokenization
    lowercase = text.lower()
    words = word_tokenize(lowercase)
    print("\n2. TOKENIZATION:")
    print(words)

    # 3. Stopword Removal
    filtered = [
        word for word in words
        if word.isalpha() and word not in stop_words
    ]
    print("\n3. STOPWORD REMOVAL:")
    print(filtered)

    # 4. Punctuation Removal
    no_punctuation = re.sub(r"[^\w\s]", "", lowercase)
    print("\n4. PUNCTUATION REMOVAL:")
    print(no_punctuation)

    # 5. Stemming + Lemmatization
    clean_words = word_tokenize(no_punctuation)

    stems = [stemmer.stem(word) for word in clean_words]
    lemmas = [lemmatizer.lemmatize(word) for word in clean_words]

    print("\n5. STEMMING:")
    print(stems)

    print("\n   LEMMATIZATION:")
    print(lemmas)
```

## Reflection

### Why preprocessing matters

1. **Tokenization** breaks text into smaller units such as words and sentences,
   making raw language easier for NLP models to process.

2. **Stopword removal** reduces common words that may add little useful
   information and can reduce vocabulary size.

3. **Punctuation removal** reduces unnecessary tokens and noise during
   downstream text processing.

4. **Stemming** reduces words to shorter root-like forms. It is fast but can
   produce words that are not valid dictionary words.

5. **Lemmatization** converts words toward meaningful dictionary forms and is
   generally more linguistically useful than aggressive stemming.

### What can break if preprocessing is skipped?

Without preprocessing, inconsistent casing, punctuation, stopwords, and
different word forms can increase noise and vocabulary size. This can make
NLP pipelines less efficient and may reduce model performance.
